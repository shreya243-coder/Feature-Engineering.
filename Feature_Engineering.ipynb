##Feature Engineering


**1. What is a parameter?**  
A parameter is a model-internal variable that is learned from training data, such as weights in linear regression. Parameters help the model make accurate predictions.

---

**2. What is correlation?**  
Correlation measures the strength and direction of a linear relationship between two variables. It ranges from -1 to 1.

---

**3. What does negative correlation mean?**  
A negative correlation means as one variable increases, the other decreases. For example, more exercise might be correlated with lower body fat.

---

**4. Define Machine Learning. What are the main components in Machine Learning?**  
Machine Learning is a method where algorithms learn patterns from data to make predictions or decisions. Main components: data, features, model, loss function, optimizer, and evaluation metric.

---

**5. How does loss value help in determining whether the model is good or not?**  
The loss value quantifies how far the model’s predictions are from the actual values. A lower loss indicates better model performance.

---

**6. What are continuous and categorical variables?**  
Continuous variables are numeric (e.g., age, salary). Categorical variables are labels or groups (e.g., gender, color).

---

**7. How do we handle categorical variables in Machine Learning? What are the common techniques?**  
Common techniques: Label Encoding (assigning numbers to categories) and One-Hot Encoding (creating binary columns). These make categories usable by ML models.

---

**8. What do you mean by training and testing a dataset?**  
Training dataset is used to teach the model, while the testing dataset is used to evaluate how well the model generalizes to unseen data.

---

**9. What is sklearn.preprocessing?**  
It’s a module in Scikit-learn that provides functions for scaling, encoding, normalizing, and transforming data before modeling.

---

**10. What is a Test set?**  
A test set is a portion of the dataset used to evaluate the model’s performance on unseen data after training.

**17. What is causation? Explain difference between correlation and causation with an example.**  
Causation means one variable directly affects another.  
**Correlation**: Ice cream sales ↑ and drowning ↑ (both due to summer).  
**Causation**: Increased speed causes higher braking distance.

**18. What is an Optimizer? What are different types of optimizers? Explain each with an example.**  
Optimizers adjust model parameters to minimize the loss.  
Types:  
- **SGD**: Basic, updates after each sample.  
- **Adam**: Adapts learning rate, faster convergence.  
- **RMSprop**: Handles learning rate for each parameter individually.
